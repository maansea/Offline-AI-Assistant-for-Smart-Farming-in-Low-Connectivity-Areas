```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Plot settings
plt.rcParams['figure.figsize'] = (12, 6)
sns.set_style('whitegrid')
sns.set_palette('husl')

print('Libraries loaded ✅')
```

## 1. Load Datasets

```python
# Load crop dataset
crop_df = pd.read_csv('../data/raw/crop_recommendation.csv')
print('Crop Dataset Shape:', crop_df.shape)
crop_df.head()
```

```python
# Load fertilizer dataset
fert_df = pd.read_csv('../data/raw/fertilizer_recommendation.csv')
print('Fertilizer Dataset Shape:', fert_df.shape)
fert_df.head()
```

## 2. Basic Info & Missing Values

```python
print('=== CROP DATASET INFO ===')
print(crop_df.info())
print('\nMissing Values:')
print(crop_df.isnull().sum())
```

```python
print('=== FERTILIZER DATASET INFO ===')
print(fert_df.info())
print('\nMissing Values:')
print(fert_df.isnull().sum())
```

## 3. Statistical Summary

```python
print('=== CROP DATASET STATISTICS ===')
crop_df.describe()
```

```python
print('=== FERTILIZER DATASET STATISTICS ===')
fert_df.describe()
```

## 4. Target Class Distribution

```python
# Crop distribution
plt.figure(figsize=(14, 6))
crop_counts = crop_df['label'].value_counts()
sns.barplot(x=crop_counts.index, y=crop_counts.values, palette='husl')
plt.title('Crop Distribution in Dataset', fontsize=16)
plt.xlabel('Crop')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('../data/processed/crop_distribution.png', dpi=150)
plt.show()
```

```python
# Fertilizer distribution
plt.figure(figsize=(12, 5))
fert_counts = fert_df['Fertilizer Name'].value_counts()
sns.barplot(x=fert_counts.index, y=fert_counts.values, palette='Set2')
plt.title('Fertilizer Distribution in Dataset', fontsize=16)
plt.xlabel('Fertilizer')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('../data/processed/fertilizer_distribution.png', dpi=150)
plt.show()
```

## 5. Feature Distributions

```python
# Crop feature distributions
feature_cols = ['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']
fig, axes = plt.subplots(2, 4, figsize=(18, 10))
axes = axes.flatten()

for i, col in enumerate(feature_cols):
    axes[i].hist(crop_df[col], bins=30, color='steelblue', edgecolor='white')
    axes[i].set_title(f'Distribution of {col}', fontsize=12)
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frequency')

axes[-1].axis('off')
plt.suptitle('Crop Dataset - Feature Distributions', fontsize=16)
plt.tight_layout()
plt.savefig('../data/processed/crop_feature_distributions.png', dpi=150)
plt.show()
```

## 6. Correlation Heatmap

```python
# Crop correlation heatmap
plt.figure(figsize=(10, 8))
corr = crop_df[feature_cols].corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm',
            square=True, linewidths=0.5)
plt.title('Crop Dataset - Feature Correlation Heatmap', fontsize=16)
plt.tight_layout()
plt.savefig('../data/processed/crop_correlation_heatmap.png', dpi=150)
plt.show()
```

## 7. Feature vs Target (Boxplots)

```python
# Nitrogen levels per crop
plt.figure(figsize=(16, 6))
sns.boxplot(data=crop_df, x='label', y='N', palette='husl')
plt.title('Nitrogen (N) Levels per Crop', fontsize=16)
plt.xlabel('Crop')
plt.ylabel('Nitrogen Level')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('../data/processed/nitrogen_per_crop.png', dpi=150)
plt.show()
```

```python
# Rainfall levels per crop
plt.figure(figsize=(16, 6))
sns.boxplot(data=crop_df, x='label', y='rainfall', palette='Set3')
plt.title('Rainfall Levels per Crop', fontsize=16)
plt.xlabel('Crop')
plt.ylabel('Rainfall (mm)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('../data/processed/rainfall_per_crop.png', dpi=150)
plt.show()
```

## 8. Summary & Key Findings

```python
print('=== EDA SUMMARY ===')
print(f'Crop Dataset: {crop_df.shape[0]} rows, {crop_df.shape[1]} columns')
print(f'Unique Crops: {crop_df["label"].nunique()}')
print(f'Missing Values: {crop_df.isnull().sum().sum()}')
print()
print(f'Fertilizer Dataset: {fert_df.shape[0]} rows, {fert_df.shape[1]} columns')
print(f'Unique Fertilizers: {fert_df["Fertilizer Name"].nunique()}')
print(f'Missing Values: {fert_df.isnull().sum().sum()}')
print()
print('✅ EDA Complete! Proceed to preprocessing and model training.')
```
